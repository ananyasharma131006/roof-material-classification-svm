```python
import joblib
import os

MODEL_PATH = f"{ROOT}/features/svm_custom_4class_final.joblib"

print("Model exists:", os.path.exists(MODEL_PATH))

model = joblib.load(MODEL_PATH)

print("\nModel type:")
print(type(model))

print("\nModel classes:")
print(model.classes_)

print("\nModel parameters:")
print(model.get_params())

print("\nModel feature count:")
print(model.n_features_in_)

print("\nFeature names:")
print(getattr(model, "feature_names_in_", "Not available"))
```

```python
import pandas as pd

features = pd.read_csv(
    f"{ROOT}/features/roof_features_4class.csv"
)

validation = pd.read_csv(
    f"{ROOT}/features/svm_tuned_validation_predictions.csv"
)

print("Feature dataset:", len(features))
print("Validation predictions:", len(validation))

print("\nValidation actual class distribution:")
print(validation["actual"].value_counts())

print("\nValidation predicted class distribution:")
print(validation["predicted"].value_counts())

print("\nDuplicate validation IDs:",
      validation.duplicated(["building_id", "region"]).sum())

print("\nActual labels match feature dataset:",
      validation.merge(
          features[["building_id", "region", "roof_material"]],
          on=["building_id", "region"]
      )["actual"].eq(
          validation.merge(
              features[["building_id", "region", "roof_material"]],
              on=["building_id", "region"]
          )["roof_material"]
      ).all())
```

```python
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Load feature dataset and labelled validation records
features = pd.read_csv(f"{ROOT}/features/roof_features_4class.csv")

validation = pd.read_csv(
    f"{ROOT}/features/svm_tuned_validation_predictions.csv"
)

# Get the exact 44 feature columns expected by your saved model
feature_names = model.named_steps["scaler"].feature_names_in_.tolist()

# Match validation buildings with their feature records
df = validation[["building_id", "region", "actual"]].merge(
    features,
    on=["building_id", "region"],
    how="left",
    validate="one_to_one"
)

# Verify records and labels
assert len(df) == len(validation), "Validation rows were lost."
assert df["roof_material"].notna().all(), "Missing feature records."
assert (df["actual"] == df["roof_material"]).all(), "Actual labels mismatch."

# Prepare model input
X_val = df[feature_names]

assert X_val.shape[1] == 44
assert not X_val.isna().any().any(), "Missing feature values found."

# Predict using your actual saved final model
predictions = model.predict(X_val)

# Create output CSV
results = df[["building_id", "region", "actual"]].copy()
results["predicted"] = predictions
results["correct"] = results["actual"] == results["predicted"]
results["result"] = results["correct"].map({
    True: "Correct",
    False: "Incorrect"
})

# Save prediction CSV
output_path = f"{ROOT}/features/final_model_validation_predictions.csv"
results.to_csv(output_path, index=False)

# Summary
correct = int(results["correct"].sum())
incorrect = int((~results["correct"]).sum())
total = len(results)

print("========== FINAL MODEL VALIDATION ==========")
print("Total predictions:", total)
print("Correct predictions:", correct)
print("Incorrect predictions:", incorrect)
print("Accuracy:", round(correct / total * 100, 2), "%")

print("\nBalanced Accuracy:",
      round(balanced_accuracy_score(results["actual"], predictions) * 100, 2), "%")

print("Macro F1:",
      round(f1_score(results["actual"], predictions, average="macro") * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(results["actual"], predictions, zero_division=0))

print("\nCSV saved at:", output_path)

# Preview
display(results.head(10))
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import pandas as pd

results = pd.read_csv(
    f"{ROOT}/features/final_model_validation_predictions.csv"
)

class_names = [
    "concrete_cement",
    "healthy_metal",
    "incomplete",
    "irregular_metal"
]

print("CLASSIFICATION REPORT")
print(classification_report(
    results["actual"],
    results["predicted"],
    labels=class_names,
    target_names=class_names,
    zero_division=0
))

print("\nCONFUSION MATRIX")

cm = confusion_matrix(
    results["actual"],
    results["predicted"],
    labels=class_names
)

display(pd.DataFrame(
    cm,
    index=class_names,
    columns=class_names
))
```
